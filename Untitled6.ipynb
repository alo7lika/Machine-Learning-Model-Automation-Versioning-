```python
import os
import joblib
import json
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# 🔹 Load dataset (Example: Iris dataset from sklearn)
from sklearn.datasets import load_iris
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = iris.target  # Target variable

# 🔹 Data Preprocessing
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # Normalize features

# 🔹 Split Data into Train & Test Sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# 🔹 Train a Random Forest Classifier
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# 🔹 Model Evaluation
y_pred = model.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
report = classification_report(y_test, y_pred, target_names=iris.target_names)

print(f"✅ Model Accuracy: {accuracy:.4f}")
print(report)

# 🔹 Model Versioning (Auto-Increment)
MODEL_DIR = "models"
if not os.path.exists(MODEL_DIR):
    os.makedirs(MODEL_DIR)

# Check existing model versions
existing_models = [f for f in os.listdir(MODEL_DIR) if f.endswith('.pkl')]
latest_version = max([int(f.split('_')[-1].split('.')[0]) for f in existing_models], default=0)
new_version = latest_version + 1

# Save Model & Metadata
model_filename = f"{MODEL_DIR}/model_v{new_version}.pkl"
metadata_filename = f"{MODEL_DIR}/metadata.json"

joblib.dump(model, model_filename)
print(f"📌 Model saved as: {model_filename}")

# Save metadata
metadata = {
    "version": new_version,
    "accuracy": accuracy,
    "model_name": "RandomForestClassifier",
    "features": iris.feature_names,
}
with open(metadata_filename, "w") as f:
    json.dump(metadata, f, indent=4)

print(f"📌 Model metadata updated: {metadata_filename}")
```
